**1. Multi-Step Probability Simulator (Markov Chains)**
Used for: A state changes step-by-step based on probabilities

**Example Questions:**
**a) Board Games:** 
A token starts at square 0 on a board. You flip a coin.
Heads moves you forward 2 spaces, tails moves you backward 1 space.
What is the chance you land exactly on square 10 within 6 flips?

```python
import random

def simulate_board_game(trials=100000):
    successful_games = 0
    
    for _ in range(trials):
        position = 0
        reached_target = False

        # max of 6 flips
        for flip in range(6):  
            # Roll for probability, 50% of heads & tails
            if random.random() < 0.50:
                position += 2
            else:
                position -= 1
                
            if position == 10:
                reached_target = True
                break  # Early exit if target is hit
                
        if reached_target:
            successful_games += 1

    probability = successful_games/trials
    print(f"Successful games: {successful_games}, Simulated Probability: {probability:.4f} ")
    
simulate_board_game()
```

**b) Stoke/Value Drift:** 
A stock price starts at £50.
Every hour, it has a **70% chance to go up by £1** and a **30% chance to drop by £2**.
What is the probability that the **stock hits £45 before it hits £55**.
        

```python
import random

def simulate_stock_price(trials=100000):
    count_45 = 0
    
    for _ in range(trials):
        price = 50
        reached_45 = False
        reached_55 = False

        # suitable number of hours for price to be hit
        for hour in range(1000):  
            # Probability of price fluctuation, 0.70 for +1, 0.30 for -2
            if random.random() < 0.70:
                price += 1
            else:
                price -= 2
                
            if price == 45:
                reached_45 = True
                break  # Early exit if either target is hit
                
            if price == 55:
                reached_55 = True
                break  
                
        if reached_45 == True:
            count_45 += 1

    probability = count_45/trials
    print(f"Number of times £45 was reached first: {count_45}, Simulated Probability: {probability:.4f} ")
    
simulate_stock_price()
```

**c) Inventory/Resource Lifespan:** 
A machine has **3 battery charges**.
Each day of operation, there is a **20% chance it uses 2 charges**, a **50% chance it uses 1 charge**, and a **30% chance it uses 0 charges**. What is the probability it runs out of battery on **exactly day 3**?

```python
import random

def simulate_battery_charge(trials=100000):
    count_day3 = 0
    
    for _ in range(trials):
        battery_charge = 3
        battery_out = False

        # loop for up to 3 days
        for day in range(3):  
            # Probability of charge usage: 0.20 for -2, 0.5 for -1, 0.3 for 0
            battery_usage = random.random() # rolls once per day
            if battery_usage < 0.50: # 0.0 to 0.50
                battery_charge -= 1
            elif battery_usage < 0.70: # 0.50 to 0.70
                battery_charge -= 2
            # else nothing happens
                
            if battery_charge <= 0:
                if day == 2:
                    count_day3 += 1
                break # Stops simulating this trial
                
        if battery_out == True and day == 2:
            count_day3 += 1

    probability = count_day3 / trials
    print(f"Number of times it ran out of battery on day 3: {count_day3}, Simulated Probability: {probability:.4f} ")
    
simulate_battery_charge()
```
